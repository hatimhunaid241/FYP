```python
#Saves the parquet to a dataframe titled products
import pandas as pd

products = pd.read_parquet('../data/processed/products.parquet')

print(products.columns)
```

```python
keywords = products["keyword_source"].unique()

for keyword in keywords:
    count = len(products[products["keyword_source"] == keyword])
    print(f"For the keyword {keyword} there are {count} products")
```
